```python
# 실습 3
torch.manual_seed(0)
scores = torch.randint(50, 101, (30, 3))
print(scores.shape)
print(scores[:5])
```

```python
# 실습 3
print('1번 학생 : ', scores[0])
print('전체 학생의 첫 과목 점수 :', scores[:,0])
print('1번 학생의 두 번째 과목 점수 : ', scores[0,1])
print('마지막 학생의 점수 : ', scores[-1])
```

```python
# 실습 3
print("예제 3-1. 10~20번째 학생의 첫 번째 과목 점수")
print(f"{scores[9:20 ,0]}")
print("예제 3-2. 전체 학생의 세 번쨰 과목")
print(f"{scores[:,2]}")
print("예제 3-3. 앞 10명의 두 번째 과목")
print(f"{scores[:10, 1]}")
print("예제 3-4. 마지막 3명의 전체 점수")
print(f"{scores[-3:, :]}")
```

```python
# 실습 4
print('현재 dtype : ', scores.dtype)
scores_f = scores.float()
print(f"변환 후 : {scores_f.dtype}")
print(f"평균 : {scores_f.mean()}")
```

```python
# 실습 5
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
x = scores_f.to(device)
print(f"device : {device}")
print(f"tensor device : {x.device}")
```
